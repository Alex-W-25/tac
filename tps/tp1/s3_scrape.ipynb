# Scraping: récupération des PDF's du sous-ensemble CaMille

Dans ce notebook, nous créons un robot qui va ouvrir la page d'accueil du site https://max.de.wilde.web.ulb.be/camille/ et récupérer les pdfs présents afin de les stocker dans un fichier csv.

## Imports

```python
import os
import re
import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Récupération des pdfs



```python
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/39.0.2171.95 Safari/537.36'}

pdf = []

root_url = f"https://max.de.wilde.web.ulb.be/camille/"
response = requests.get(root_url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
for link in soup.find_all("a"):
    href = link.get('href')
    title = link.text.strip()
    url = root_url + href
    pdf.append([url, title])
```

## Création d'un dataframe avec les pdfs


```python
df = pd.DataFrame(pdf, columns=['link', 'title'])
df
```

```python
# Sauvegarde du dataframe dans un fichier csv
df.to_csv(f"data/camille_{time.strftime('%Y%m%d')}.csv", index=False)
```

## Téléchargement des fichiers PDF

```python
# Créer un répertoire PDF s'il n'existe pas encore
pdf_path = './data/pdf'
if not os.path.exists(pdf_path):
    os.mkdir(pdf_path)
```

```python
# Récupérer les fichiers pdf un à un
url = "https://max.de.wilde.web.ulb.be/camille/KB_JB729_1939-10-31_01-00006.pdf"
filename = "KB_JB729_1939-10-31_01-00006.pdf"
# Récupérer le fichier pdf dans la RAM
response = requests.get(url)
```

```python
# Ecrire le fichier sur le disque
with open(os.path.join(pdf_path, filename), 'wb') as f:
    f.write(response.content)
```
