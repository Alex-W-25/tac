## Imports

```python
import json
import os
import requests
```

## Utiliser [OCR.space](https://ocr.space/ocrapi) pour faire de la reconnaissance de caractère sur les pdf

```python
pdf_path = './data/pdf'
filename = "KB_JB729_1939-10-31_01-00006.pdf"
way_pdf = os.path.join(pdf_path, filename)

url_api = "https://ocr.space"
payload = {
        'apikey': 'ed3f6016b888957', 
        'language': 'fre',        
    }
try:
        with open(way_pdf, 'rb') as f:
            fichiers = {'file': f}
            response = requests.post(url_api, data=payload, files=fichiers)
            resultat = response.json()
            
        if resultat.get("ParsedResults"):
            texte_extrait = resultat["ParsedResults"][0]["ParsedText"]
            print("Texte détecté :\n", texte_extrait)
        else:
            print("Erreur ou aucun texte trouvé :", resultat.get("ErrorMessage"))
            
```
